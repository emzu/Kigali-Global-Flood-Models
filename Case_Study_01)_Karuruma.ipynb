##	Case Study 1: Satellite-Derived Flood Extents for Model Calibration in Karuruma

```python
import sys
sys.path.append('...')

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

plt.style.use('publication.mplstyle')
```

# Nyabugogo Catchment


```python
df = pd.read_csv('Hydrology Station Data\Obs_simul_flow_Karuruma.csv')
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)
```

```python
fig, ax = plt.subplots(figsize=(8, 4))

# Plot directly
df.plot(y='Discharge Observed', ax=ax, label='Observed', color='blue')
df.plot(y='Discharge Simulated', ax=ax, label='Simulated', color='orange')
plt.gcf().autofmt_xdate()
plt.legend()
plt.xlabel('Date')
plt.ylabel('Discharge (m³/s)')

plt.tight_layout()
plt.show()
```
